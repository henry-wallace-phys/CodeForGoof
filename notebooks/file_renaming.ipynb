## What does this notebook do?
This notebook provides a way to rename all files in a folder (and its subfolders). By default this replaces all " " with "_"

## How to use:
There are two ways to use "rename stuff"
1. `rename_stuff()` with no arguments. This will pop up a box asking you to enter a directory path. Enter a path to the directory 
2. `rename_stuff("path/to/folder")`. This does exactly the same thing but you can see the directory path!

To run simply press `Run All` at the top of the screen (or shift+enter)

## Additional Functionality
1. You can provide the rename stuff with other strings to replace for example:
`rename_stuff(old_str="txt", new_str="xml")`. This will replace all instances of `txt` in your files with `xml`. This also works with the manual path method i.e. `rename_stuff(path/to/folder, old_str="txt", new_str="xml")`

## NOTE:
When using the "manual" method or passing different renaming things remember to enclose them in quotation marks. So `rename_stuff("path/to/folder")` will work but `rename_stuff(path/to/folder)` won't!
