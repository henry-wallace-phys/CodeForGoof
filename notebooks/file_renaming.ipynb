## What does this notebook do?
This notebook provides a way to rename all files in a folder (and its subfolders). By default this replaces all " " with "_"

## How to use:
There are two ways to use "rename stuff"
First make the renamer with `renamer = FileRenamer()`

1. Now you can call `renamer.rename_files()` with no arguments. This will pop up a box asking you to enter a directory path. Enter a path to the directory 
2. `renamer.rename_files("path/to/folder")`. This does exactly the same thing but you can see the directory path!


To run simply press `run` (the > button) on the first 2 cells below this (or shift+enter)

To undo the changes run the `renamer.undo_rename()`

## Additional Functionality
1. You can provide the rename stuff with other strings to replace for example:
`FileRenamer(old_str="txt", new_str="xml")`. This will replace all instances of `txt` in your files with `xml`. This also works with the manual path method i.e. `rename_stuff(path/to/folder, old_str="txt", new_str="xml")`

## NOTE:
When using the "manual" method or passing different renaming things remember to enclose them in quotation marks. So `rename_stuff("path/to/folder")` will work but `FileRenamer(path/to/folder)` won't!
